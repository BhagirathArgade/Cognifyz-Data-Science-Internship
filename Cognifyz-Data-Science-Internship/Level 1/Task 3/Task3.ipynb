# Task 3 (Level 1) : GEOSPATIAL ANALYSIS

-           Loading dataset

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("dataset_level1_cleaned.csv")
df.head()
```

-           Scatter Plot of Restaurant Locations

This scatter plot visualizes restaurant locations using latitude and longitude.
Each point on the plot represents a restaurant.
Clusters in the plot indicate areas where many restaurants are closely located.

```python
plt.figure(figsize=(10,6))
plt.scatter(df['Longitude'], df['Latitude'], s=10)
plt.title("Restaurant Locations (Longitude vs Latitude)")
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.grid(True)
plt.show()
```

-           Restaurant Distribution by City

This step shows how restaurants are distributed across cities.
Cities with the highest bar represent the most restaurant-dense areas.
This helps identify major food hubs or markets.

```python
city_counts = df['City'].value_counts().head(20)
city_counts
```

Bar chart for top 20 cities

```python
plt.figure(figsize=(10,6))
city_counts.plot(kind='bar')
plt.title("Top 20 Cities with Most Restaurants")
plt.xlabel("City")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

-           Restaurant Distribution by Country

This shows how restaurants are distributed across different countries.
Some countries in the dataset have very high counts due to larger cities being included.

```python
country_counts = df['Country Code'].value_counts()
country_counts
```

Bar Chart

```python
plt.figure(figsize=(8,5))
country_counts.plot(kind='bar', color='orange')
plt.title("Restaurant Count by Country")
plt.xlabel("Country Code")
plt.ylabel("Number of Restaurants")
plt.show()
```

-           Correlation: Location vs Rating

Correlation analysis checks whether restaurant ratings depend on the restaurant’s physical location.
The correlation matrix usually shows values close to 0, meaning:

Ratings have no strong relationship with latitude

Ratings have no strong relationship with longitude

Scatter plots confirm this — points appear randomly spread with no upward or downward trend.

This means restaurant quality is not determined by geographic coordinates.

We check statistical correlation first.

```python
df[['Latitude', 'Longitude', 'Aggregate rating']].corr()
```

Scatter plot: Latitude vs Rating

```python
plt.scatter(df['Latitude'], df['Aggregate rating'], s=10)
plt.title("Latitude vs Rating")
plt.xlabel("Latitude")
plt.ylabel("Aggregate Rating")
plt.show()
```

Scatter plot: Longitude vs Rating

```python
plt.scatter(df['Longitude'], df['Aggregate rating'], s=10)
plt.title("Longitude vs Rating")
plt.xlabel("Longitude")
plt.ylabel("Aggregate Rating")
plt.show()
```
