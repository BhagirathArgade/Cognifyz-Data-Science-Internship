# Task 3 (level 2) : Price Range vs. Rating Analysis

-           Load the dataset

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("dataset_level1_cleaned.csv")
df.head()

        
```

-           Average Rating per Price Range

This calculates the mean rating for each price range (1 to 4).
It tells us whether expensive restaurants perform better in customer satisfaction.

```python
price_rating = df.groupby('Price range')['Aggregate rating'].mean()
price_rating
```

Bar Plot :


This bar chart visually shows how ratings differ across price groups.
A rising pattern would indicate that higher-priced restaurants generally have higher ratings.

```python
plt.figure(figsize=(7,5))
price_rating.plot(kind='bar', color='skyblue')
plt.title("Average Rating by Price Range")
plt.xlabel("Price Range")
plt.ylabel("Average Aggregate Rating")
plt.show()
```

-           Count Restaurants in Each Price Range

```python
price_counts = df['Price range'].value_counts().sort_index()
price_counts
```

Plot it :

Shows how many restaurants fall into each price category.
This helps us understand whether some categories dominate the dataset.

```python
plt.figure(figsize=(7,5))
price_counts.plot(kind='bar', color='orange')
plt.title("Restaurant Count by Price Range")
plt.xlabel("Price Range")
plt.ylabel("Number of Restaurants")
plt.show()
```

-           Boxplot: Distribution of Ratings in Each Price Category

The boxplot shows medians, quartiles, and outliers.
You can visually compare which price range has the best rating distribution.

```python
plt.figure(figsize=(8,6))
df.boxplot(column='Aggregate rating', by='Price range')
plt.title("Rating Distribution by Price Range")
plt.suptitle("")
plt.xlabel("Price Range")
plt.ylabel("Aggregate Rating")
plt.show()
```

-           Correlation Between Price & Rating

This gives a numerical correlation value.
If the number is positive (e.g., 0.2 or 0.3), then higher price range has a slight positive relationship with higher ratings.
If near zero, relationship is weak.

```python
df[['Price range', 'Aggregate rating']].corr()
```
